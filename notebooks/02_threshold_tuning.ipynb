# PIDE: Threshold Tuning for L2 Semantic Layer
This notebook explores the distribution of cosine similarities to select the optimal $\tau$ (tau) threshold.

```python
import json
import numpy as np
import matplotlib.pyplot as plt
from sentence_transformers import SentenceTransformer
import faiss

# 1. Load Exemplars
with open('../data/attack_exemplars.json', 'r') as f:
    data = json.load(f)
exemplars = data['exemplars']

# 2. Load Model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 3. Compute distribution of internal similarities (Cross-validation style)
embeddings = model.encode(exemplars)
faiss.normalize_L2(embeddings)
sim_matrix = np.dot(embeddings, embeddings.T)

plt.hist(sim_matrix.flatten(), bins=50)
plt.title("Similarity Distribution within Attack Exemplars")
plt.show()
```

### Precision/Recall vs Threshold
By varying the threshold, we can find the 'elbow' where recall remains high but false positives are minimized.
